# Datos de ropa
Preguntas 17, 18, 20 y 21. Se toma en cuenta la columna `cantidad`.

## Cargar librerías

```python
import numpy as np
import pandas as pd
```

## Cargar datos

```python
url = 'https://raw.githubusercontent.com/rpizarrog/probabilidad-y-estad-stica/master/datos/datos_ropa.csv'
datos = pd.read_csv(url)
datos['tallas'] = datos['tallas'].astype(str)
print(datos.shape)
print(datos.head(42))
print(datos.tail(42))
```

## Espacio muestral (total de prendas)

```python
total = datos['cantidad'].sum()
print('Total de prendas:', total)
```

## Tabla resumen por producto

```python
resumen = datos.groupby('productos')['cantidad'].sum().to_frame('prendas')
resumen['probabilidad_%'] = (resumen['prendas']/total*100).round(2)
resumen
```

## Pregunta 17: P(zapatos)

```python
zapatos = datos[datos['productos'] == 'ZAPATO']
print('Zapatos:', zapatos['cantidad'].sum())
print('P(zapatos) =', round(zapatos['cantidad'].sum()/total*100, 2), '%')
```

## Pregunta 18: P(zapatos número 24)

```python
z24 = zapatos[zapatos['tallas'] == '24']
print('Zapatos talla 24:', z24['cantidad'].sum())
print('P(zapato 24) =', round(z24['cantidad'].sum()/total*100, 2), '%')
```

## Pregunta 20: P(zapatos femenino número 24)

```python
z24f = z24[z24['generos'] == 'FEMENINO']
print('Zapatos femenino talla 24:', z24f['cantidad'].sum())
print('P =', round(z24f['cantidad'].sum()/total*100, 2), '%')
```

## Pregunta 21: P(pantalón femenino marca X)

```python
pfx = datos[(datos['productos'] == 'PANTALON') & (datos['generos'] == 'FEMENINO') & (datos['marcas'] == 'X')]
print('Pantalones femeninos marca X:', pfx['cantidad'].sum())
print('P =', round(pfx['cantidad'].sum()/total*100, 2), '%')
```
