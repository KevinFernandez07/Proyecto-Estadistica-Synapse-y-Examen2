# Alumnos y deportes
Preguntas 6, 7 y 22 del cuestionario.

## Cargar librerías

```python
import numpy as np
import pandas as pd
```

## Cargar datos

```python
url = 'https://raw.githubusercontent.com/rpizarrog/probabilidad-y-estad-stica/master/2024/datos/datos_alumnos_deportes.csv'
datos = pd.read_csv(url)
print(datos.shape)
datos.head(10)
```

## Pregunta 6: P(sexo = F)

```python
total = len(datos)
femeninos = set(datos[datos['sexo'] == 'F']['nombres'])
print('Mujeres:', len(femeninos), 'de', total)
print('P(F) =', round(len(femeninos)/total*100, 2), '%')
```

## Pregunta 7: P(fútbol ∩ básquetbol)

```python
futbol = set(datos[datos['futbol'] == 1]['nombres'])
basquetbol = set(datos[datos['basquetbol'] == 1]['nombres'])
ambos = futbol & basquetbol
print('Juegan fútbol:', len(futbol))
print('Juegan básquetbol:', len(basquetbol))
print('Juegan ambos:', len(ambos))
print('P(fútbol ∩ básquetbol) =', round(len(ambos)/total*100, 2), '%')
```

## Pregunta 22: P(masculino | básquetbol)
P(M | B) = P(M ∩ B) / P(B)

```python
masculino = set(datos[datos['sexo'] == 'M']['nombres'])
m_y_b = masculino & basquetbol

p_m_y_b = len(m_y_b) / total
p_b = len(basquetbol) / total

print('P(M ∩ B) =', round(p_m_y_b*100, 2), '%')
print('P(B)     =', round(p_b*100, 2), '%')
print('P(M | B) =', round(p_m_y_b/p_b*100, 2), '%')
```

**Nota:** el valor correcto de P(M | B) es ≈ 40.54 %, que no aparece en las opciones. El 15 % es la intersección P(M ∩ B), sin dividir entre P(B).
