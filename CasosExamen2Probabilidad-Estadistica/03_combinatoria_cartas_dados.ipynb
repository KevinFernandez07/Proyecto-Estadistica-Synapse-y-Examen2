## Cargar librerías

```python
import math as mt
import numpy as np
import pandas as pd
from itertools import combinations, product
```

## Pregunta 3: combinaciones de 3 sin repetición (7 elementos)

```python
caracteres7 = ['A', 'B', 'C', '0', '1', '2', '3']
n, r = 7, 3
print('Fórmula:', mt.factorial(n) / (mt.factorial(r) * mt.factorial(n - r)))
comb7 = list(combinations(caracteres7, r))
print('Con itertools:', len(comb7))
```

## Preguntas 8 y 23: baraja del 1 al 7 (A, B, C, D), combinaciones de 2 en 2

```python
def crear_baraja(k):
    return [f'{p}{i}' for p in 'ABCD' for i in range(1, k + 1)]

def tabla_cartas(k):
    baraja = crear_baraja(k)
    filas = []
    for c1, c2 in combinations(baraja, 2):
        filas.append({'carta1': c1, 'carta2': c2, 'suma': int(c1[1:]) + int(c2[1:])})
    return pd.DataFrame(filas)

tabla7 = tabla_cartas(7)
total7 = len(tabla7)
print('Total de combinaciones:', total7)
tabla7.head(10)
```

```python
menor_igual_10 = (tabla7['suma'] <= 10).sum()
igual_10 = (tabla7['suma'] == 10).sum()
print('Suma <= 10:', menor_igual_10, 'de', total7, '->', round(menor_igual_10/total7*100, 2), '%   (Pregunta 8)')
print('Suma == 10:', igual_10, 'de', total7, '->', round(igual_10/total7*100, 2), '%   (Pregunta 23)')
```

## Pregunta 10: cuántas veces aparece la 'C' (10 caracteres, grupos de 3)
Ojo: el enunciado dice n = 7, pero la lista tiene 10 elementos, y las opciones (32 a 38) solo cuadran con los 10.

```python
caracteres10 = ['A', 'B', 'C', 'D', '0', '1', '2', '3', '4', '5']
comb10_3 = list(combinations(caracteres10, 3))
veces_C = sum(1 for c in comb10_3 if 'C' in c)
print('Total de combinaciones:', len(comb10_3))
print("La 'C' aparece en", veces_C, 'combinaciones')
```

## Pregunta 12: parejas Hombre-Mujer

```python
personas = ['H JUAN', 'M PATY', 'M SANDRA', 'H HECTOR', 'M MARIA', 'H, MARIO', 'H GERARDO', 'M SANDRA',
            'H ERNESTO', 'H RUBEN', 'M PATY', 'M LUCY', 'H LUIS', 'H ANTONIO', 'H CARLOS', 'M GABY']

parejas = list(combinations(personas, 2))
mixtas = [p for p in parejas if p[0][0] != p[1][0]]
print('Total de parejas:', len(parejas))
print('Hombres:', sum(p[0] == 'H' for p in personas), '| Mujeres:', sum(p[0] == 'M' for p in personas))
print('Parejas Hombre-Mujer:', len(mixtas))
```

## Preguntas 13 y 16: dos dados (permutaciones con repetición)

```python
dados = pd.DataFrame(list(product(range(1, 7), repeat=2)), columns=['dado1', 'dado2'])
dados['suma'] = dados['dado1'] + dados['dado2']
print('Total de posibilidades:', len(dados))
dados.head(12)
```

```python
# Pregunta 16: suma = 7
suma7 = dados[dados['suma'] == 7]
print('Suma = 7:', len(suma7))
print(suma7)

# Pregunta 13: pares (1,x) o (x,1) con x de 2 a 6
pares = dados[((dados['dado1'] == 1) & (dados['dado2'] != 1)) | ((dados['dado2'] == 1) & (dados['dado1'] != 1))]
print('\nPares 1 con 2..6 en cualquier orden:', len(pares), 'de', len(dados))
```
