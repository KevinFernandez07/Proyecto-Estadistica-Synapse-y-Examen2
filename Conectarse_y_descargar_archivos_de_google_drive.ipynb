# Importar librerías


```python
import gdown
import pandas as pd
import os
```

## Crear funciones

Funciones para este ejercicio


```python
def f_descargar_datos(ruta_google_drive):
    """
    Descarga una carpeta pública de Google Drive
    en la carpeta local 'datos'.

    Returns:
        list: Lista de archivos descargados.
    """

    try:
        archivos = gdown.download_folder(
            url=ruta_google_drive,
            output="datos",
            quiet=False
        )

        print("\nRepositorio de datos descargado.")

        return archivos

    except Exception as e:
        print(f"Error al descargar los datos: {e}")
        return None


def f_cargar_datos_csv(ruta):
  """
  Cargar archivos csv
  Returns:
    pandas.DataFrame: El archivo cargadao
  """
  try:
    datos = pd.read_csv(ruta)
    print(f"Archivo cargado: {ruta}")
    return datos
  except Exception as e:
    print(f"Error al cargar archivo {ruta}: {e}")
    return None
```

Se descargan los archivos de *google drive*. Con la instrucción gdown.download_folder() se le dice que se descarguen los archuivos y se crea una carpeta temproal aquí en google colab llamada datos que dond eestarán los datos.



```python
url_google_drive = "https://drive.google.com/drive/folders/11a7TEYwZR6Om6xLNGpU9yEf4AUcR4Myo?usp=sharing"

archivos = f_descargar_datos(url_google_drive)

```

## Cargar un archivo

Con la función *f_cargar_datos()* se carga un conjunto de datos.

```python
# Cargar archjio específico de los que se descargaron
# ejemplo:datos_educacion_100.csv de la carpeta datos
# datos/datos_educacion_100.csv

print ("Los archivos 'csv' cargados ")
print(archivos)
print()
print("Cargar un archivo específico... 'datos/datos_educacion_100.csv'")
datos = f_cargar_datos_csv("datos/datos_educacion_100.csv")
datos
```
